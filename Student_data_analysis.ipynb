```python
# Step 1: Create the scatter plot
plt.figure(figsize=(10, 6))

plt.title('UN Estimates vs World Bank Estimates')
plt.xlabel('World Bank Estimate')
plt.ylabel('UN Estimate')
plt.grid(True)

import numpy as np

# Calculate the trend line and a colour:
m, b = np.polyfit(gdp_df['WorldBank_Estimate'], gdp_df['UN_Estimate'], 1)  # Linear fit
plt.plot(gdp_df['WorldBank_Estimate'], m * gdp_df['WorldBank_Estimate'] + b, color='orange', linestyle='-')

plt.scatter(gdp_df['WorldBank_Estimate'], gdp_df['UN_Estimate'], color='green', alpha=0.6)
```

```python
# Use the exact filename you have just uploaded
stu_df = pd.read_csv('/content/drive/MyDrive/student.csv')
```

```python
#Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print(stu_df.columns.tolist()) # check column names only
```

```python
# To create a Histogram:
plt.figure(figsize=(10, 6))
plt.hist(stu_df['mark'], bins=10, edgecolor='black',color='crimson')
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
counts, bins, patches = plt.hist(stu_df['mark'], bins=10, edgecolor='black')

colors = ['red', 'blue', 'green', 'orange', 'purple', 'cyan', 'yellow', 'pink', 'brown', 'grey']
for patch, color in zip(patches, colors):
  patch.set_facecolor(color)

plt.xlabel('Mark')
plt.ylabel('Frequency')
plt.title('Marks histogram with different colours per bin')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='class', y='mark', hue='class', data=stu_df, palette='Set2')
plt.title('Marks by Class')
plt.xlabel('Class')
plt.ylabel('Marks')
plt.show()
```
